# Day Sensor Test

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
obs = pd.read_csv("../data/log.csv", parse_dates=["timestamp"])
ver = pd.read_csv("../data/weather.csv", parse_dates=["date_time"])
```

```python
obs = obs[obs["timestamp"].dt.date != pd.Timestamp("2026-10-18").date()]
```

```python
obs
```

```python
ver
```

```python
plt.plot(obs["timestamp"], obs["temperature_f"], label="sensor recording")
plt.plot(ver["date_time"], ver[" temperature"], label="weather API")

plt.title("Sensor vs. Weather API Temperature")
plt.ylabel("Temperature F")

plt.gcf().autofmt_xdate()
plt.legend()
```

```python
plt.plot(obs["timestamp"], obs["humidity_pct"], label="sensor recording")
plt.plot(ver["date_time"], ver[" humidity"], label="weather API")

plt.title("Sensor vs. Weather API Humidity")
plt.ylabel("Humidity %")

plt.gcf().autofmt_xdate()
plt.legend()
```

```python
plt.plot(obs["timestamp"], obs["pressure_hpa"], label="sensor recording")
plt.plot(ver["date_time"], ver[" pressure"], label="weather API")

plt.title("Sensor vs. Weather API Air Pressure")
plt.ylabel("Pressure (hpa)")

plt.gcf().autofmt_xdate()
plt.legend()
```

```python
plt.plot(obs["timestamp"], obs["power_mw"], label="sensor recording")

plt.title("Power Consumption")
plt.ylabel("mW")

plt.gcf().autofmt_xdate()
```
